# NB sBERT vs LaBSE: hvilken forstår nynorsk og bokmål best?

```python
from sentence_transformers import SentenceTransformer, util

labse_model = SentenceTransformer('sentence-transformers/LaBSE', device="cuda")
sbert_model = SentenceTransformer('NbAiLab/nb-sbert-base', device="cuda")
```

```python
nn = "Den eigentlege røyndommen er eit mysterium, det er ein løyndom som førebels korkje \
menneske eller dyr forstår. Me er samde i det"
bm = "Den egentlige virkeligheten er et mysterium, det er en hemmelighet som foreløpig hverken  \
mennesker eller dyr forstår. Vi er enige i det"
```

```python
labse_embs = labse_model.encode([nn, bm])
sbert_embs = sbert_model.encode([nn, bm])
```

```python
util.cos_sim(*labse_embs)
```

```python
util.cos_sim(*sbert_embs)
```
