```python
from pathlib import Path
import pandas as pd
import os
import re

def extract_dfs(domain, lang_code):
    data_path_2024 = os.environ.get("MALFRID_2024", None)
    print(data_path_2024)
    source_p = Path(data_path_2024)
    out_p = Path(f"output/{domain}_{lang_code}.json")

    if not out_p.exists():
        filer = [e for e in source_p.iterdir() if domain in e.name and lang_code in e.name and "html" in e.name]
        if filer: 
            df = []
            for e in filer:
                with e.open("rb") as f:
                    df.append(pd.read_json(e, lines=True))
            df = pd.concat(df)
            df.index = range(len(df))
            df.to_json(out_p, index=False)
    df = pd.read_json(out_p)
    return df

df_nob = extract_dfs("lanekassen", "nob")
df_nno = extract_dfs("lanekassen", "nno")
```

Fjern språkkode and sammenlign url med SequenceMatcher

```python
import re
from difflib import SequenceMatcher
from collections import defaultdict
#from Levenshtein import distance

def extract_language_code(url):
    match = re.search(r'\/([a-z]{2}-[A-Z]{2}|nynorsk)\/', url)
    if match:
        return match.group(1)
    return None

def normalize_url(url):
    language_code = extract_language_code(url)
    if language_code:
        url = re.sub(r'\/[a-z]{2}-[A-Z]{2}\/', '/xx-XX/', url)
        print(url)
    return url

def compare_urls(url1, url2):
    normalized_url1 = normalize_url(url1)
    normalized_url2 = normalize_url(url2)

    similarity_score = SequenceMatcher(None, normalized_url1, normalized_url2).ratio()
    #similarity_score = Levenshtein.distance(normalized_url1, normalized_url2, weights=None, processor=None, score_cutoff=None, score_hint=None)
    return similarity_score

def write_matches(url1, url2):
    similarity_score = compare_urls(url1, url2)
    return similarity_score

threshold = 0.98
result_dict = defaultdict(list)
num_matches=0
for nn_row in df_nno.itertuples():
    for nb_row in df_nob.itertuples():
        nn_url = getattr(nn_row, 'url')
        nb_url = getattr(nb_row, 'url')
        
        similarity_score = write_matches(nn_url, nb_url)
        
        if similarity_score >= threshold:
            num_matches += 1
            
            result_dict["nynorsk_doc_hash"].append(getattr(nn_row, 'doc_hash'))
            result_dict["bokmaal_doc_hash"].append(getattr(nb_row, 'doc_hash'))
            result_dict["nynorsk_url"].append(nn_url)
            result_dict["bokmaal_url"].append(nb_url)
            result_dict["nynorsk_fulltext"].append(getattr(nb_row, 'fulltext'))
            result_dict["bokmaal_fulltext"].append(getattr(nb_row, 'fulltext'))

result_df = pd.DataFrame(result_dict)
print(f"Matches found: {num_matches}")
```

Fjern url-er som ender med et tall for å unngå dårlige matcher (feks  https://lanekassen.no/nn-NO/presse-og-samfunnskontakt/nyheiter/fastrentene-i-lanekassen-stiger2/ https://lanekassen.no/nn-NO/presse-og-samfunnskontakt/nyheiter/fastrentene-i-lanekassen-stiger3/)

```python
def ends_with_digit(string):
    return string[-1].isdigit() if string else False

result_df_filtered = result_df[~result_df['nynorsk_url'].apply(ends_with_digit) & ~result_df['bokmaal_url'].apply(ends_with_digit)]
```
