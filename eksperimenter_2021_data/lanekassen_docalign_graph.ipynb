```python
from pathlib import Path
import pandas as pd
from hemmelig import data_path

p = Path(data_path)

filer = [e for e in p.iterdir() if "lanekassen" in e.name]

dfs = []
for e in filer:
    with e.open("rb") as f:
        dfs.append(pd.read_json(e, lines=True))


df = pd.concat(dfs)
df
```

```python
nynorske = df[df.lang == "nno"]
nynorske.index = range(len(nynorske))

bokmålske = df[df.lang == "nob"]
bokmålske.index = range(len(bokmålske))

len(nynorske), len(bokmålske)
```

```python
from sentence_transformers import SentenceTransformer, util

model = SentenceTransformer('NbAiLab/nb-sbert-base', device="cuda")
```

```python
nynorske_sents = [(e.doc_hash, sent) for e in nynorske.itertuples() for sent in e.fulltext if sent]
nynorske_sents = pd.DataFrame(nynorske_sents, columns=["doc_hash", "sentence"])
nynorske_sents = nynorske_sents.drop_duplicates("sentence")
nynorske_sents.index = range(len(nynorske_sents))
nynorske_sents
```

```python
bokmål_sents = [(e.doc_hash, sent) for e in bokmålske.itertuples() for sent in e.fulltext if sent]
bokmål_sents = pd.DataFrame(bokmål_sents, columns=["doc_hash", "sentence"])
bokmål_sents = bokmål_sents.drop_duplicates("sentence")
bokmål_sents.index = range(len(bokmål_sents))
bokmål_sents
```

```python
nynorsk_embeddings = model.encode(list(nynorske_sents.sentence))
bokmål_embeddings = model.encode(list(bokmål_sents.sentence))

search_result = util.semantic_search(nynorsk_embeddings, bokmål_embeddings, top_k=1)
threshold = 0.95
matches = [(i, e[0]) for i, e in enumerate(search_result) if e[0]["score"] > threshold]
```

```python
doc_edges = [(nynorske_sents.iloc[nynorsk_sent_i].doc_hash, bokmål_sents.iloc[match["corpus_id"]].doc_hash) for nynorsk_sent_i, match in matches]
```
